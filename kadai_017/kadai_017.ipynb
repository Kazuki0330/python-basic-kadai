```python
#kadai_017 組み合わせ
class Human:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}さんは{self.age}歳なので、大人です。")
        else:
            print(f"{self.name}さんは{self.age}歳なので、大人ではありません。")


#  複数のHumanインスタンスを生成
human_list = [
    Human("太郎", 25),
    Human("花子", 18),
    Human("次郎", 20),
    Human("三郎", 15),
]

#check_adult メソッド
for human in human_list:
    human.check_adult()
```
